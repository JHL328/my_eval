```python
from pathlib import Path
import pandas as pd

try:
    from IPython.display import display
except ImportError:  # fallback when run as a script
    display = print

def load_parquet(path_str: str) -> pd.DataFrame:
    path = Path(path_str).expanduser().resolve()
    df = pd.read_parquet(path)
    print(f'Loaded {path.name}: {len(df)} rows x {len(df.columns)} columns')
    return df

df = load_parquet(file_path_1)
display(df.head())
display(df.dtypes)
```
