```python
model_path = "/mnt/sharefs/users/haolong.jia/RL-model/sft/steel_lamb/checkpoint-5472"
query = "what is the capital of France?"

from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForCausalLM.from_pretrained(model_path, torch_dtype=torch.float16, device_map="auto")

# Apply chat template
messages = [
    {"role": "user", "content": query}
]

# Get the formatted input with chat template
formatted_input = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
print("=" * 50)
print("Formatted input after applying chat template:")
print("=" * 50)
print(formatted_input)
print("=" * 50)

# Tokenize for model input
inputs = tokenizer.apply_chat_template(messages, tokenize=True, add_generation_prompt=True, return_tensors="pt")
inputs = inputs.to(model.device)

print(f"\nTokenized input shape: {inputs.shape}")
print(f"Tokenized input: {inputs}")

# Generate response
with torch.no_grad():
    outputs = model.generate(
        inputs,
        max_new_tokens=100,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

# Decode the full output
full_response = tokenizer.decode(outputs[0], skip_special_tokens=False)
print("\n" + "=" * 50)
print("Full output (with special tokens):")
print("=" * 50)
print(full_response)

# Decode only the generated part
generated_tokens = outputs[0][inputs.shape[-1]:]
generated_text = tokenizer.decode(generated_tokens)
print("\n" + "=" * 50)
print("Generated response only:")
print("=" * 50)
print(generated_text)
```
